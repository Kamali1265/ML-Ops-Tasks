<h2 style="color:green" align="center">Machine Learning With Python: Save And Load Trained Model</h2>

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("traffic.csv")

model = LinearRegression()

model.fit(df[['vehicle_count']], df['duration'])
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
model.predict([[300]])
```

```python
import pickle

with open('traffic_model_pickle', 'wb') as file:
    pickle.dump(model, file)
```

<h3 style='color:purple'>Save Model To a File Using Python Pickle</h3>

```python
import pickle
```

```python
with open('traffic_model_pickle', 'wb') as file:
    pickle.dump(model, file)
```

<h4 style='color:purple'>Load Saved Model</h4>

```python
with open('traffic_model_pickle', 'rb') as file:
    tm = pickle.load(file)
```

```python
tm.coef_
```

```python
tm.intercept_
```

```python
tm.predict([[400]])
```

# Save Trained Model Using joblib

```python
import joblib
```

```python
joblib.dump(model, 'traffic_model_joblib')
```

<h4 style='color:purple'>Load Saved Model</h4>

```python
tm = joblib.load('traffic_model_joblib')
```

```python
tm.coef_
```

```python
tm.intercept_
```

```python
tm.predict([[500]])
```
