**LOGISTIC REGRESSION**

```python
import pandas as pd
from matplotlib import pyplot as plt

%matplotlib inline
```

```python
df = pd.read_csv("traffic_logistic.csv")
df.head()
```

```python
plt.scatter(
    df.vehicle_count,
    df.accident_risk,
    marker='*',
    color='purple'
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    df[['vehicle_count']],
    df.accident_risk,
    train_size=0.8
)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
X_test
```

```python
y_predicted = model.predict(X_test)

y_predicted
```

```python
model.predict_proba(X_test)
```

```python
model.score(X_test,y_test)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
import math

def sigmoid(x):
    return 1/(1+math.exp(-x))
```

```python
def prediction_function(vehicle_count):

    z = (
        model.coef_[0][0] * vehicle_count
        + model.intercept_[0]
    )

    y = sigmoid(z)

    return y
```

```python
prediction_function(100)
```

```python
prediction_function(300)
```

```python
prediction_function(700)
```

```python
prediction_function(650)
```
